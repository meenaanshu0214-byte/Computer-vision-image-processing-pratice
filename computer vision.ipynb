```python
import cv2
image = cv2.imread("actor.jpg")
print("Image shape:", image.shape) 
pixel = image[100, 50]
print("Pixel value (BGR) at (100, 50):", pixel)
```

```python
print("Image shape:", image.shape)
print("Image array:\n", image)
```

```python
b_channel, g_channel, r_channel = cv2.split(image)
print("Blue channel shape:", b_channel.shape)
print("Green channel shape:", g_channel.shape)
print("Red channel shape:", r_channel.shape)
merged_image = cv2.merge([b_channel, g_channel, r_channel])
cv2.imwrite("merged_image.jpg", merged_image)
cv2.imshow("Merged Image", merged_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
image[50:100, 100:200] = [255, 255, 255]
cv2.imwrite("modified_image.jpg", image)
cv2.imshow("Modified Image", image)
cv2.waitKey(10000)
cv2.destroyAllWindows()
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
_, binary_image = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)
```

```python
adaptive_image = cv2.adaptiveThreshold(gray_image, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, blockSize=11, C=2 )
plt.figure(figsize=(16, 4))
plt.subplot(1, 4, 1)
plt.title("Original Image")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.subplot(1, 4, 2)
plt.title("Grayscale Image")
plt.imshow(gray_image, cmap='gray')
plt.axis('off')
plt.subplot(1, 4, 3)
plt.title("Binary Thresholding")
plt.imshow(binary_image, cmap='gray')
plt.axis('off')
plt.subplot(1, 4, 4)
plt.title("Adaptive Thresholding")
plt.imshow(adaptive_image, cmap='gray')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('actor.jpg')
blurred = cv2.GaussianBlur(img, (5, 5), 0)
plt.subplot(1, 2, 1)
plt.title('Original')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.subplot(1, 2, 2)
plt.title('Gaussian Blurred')
plt.imshow(cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
median_blurred = cv2.medianBlur(img, 5)
plt.subplot(1, 2, 1)
plt.title('Original')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.subplot(1, 2, 2)
plt.title('Median Blurred')
plt.imshow(cv2.cvtColor(median_blurred, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
filtered = cv2.bilateralFilter(img, d=9, sigmaColor=75, sigmaSpace=75)
plt.subplot(1, 2, 1)
plt.title('Original')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.subplot(1, 2, 2)
plt.title('Bilateral Filtered')
plt.imshow(cv2.cvtColor(filtered, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
blurred = cv2.GaussianBlur(img, (5, 5), 0)

mask = cv2.subtract(img, blurred)
#
sharpened = cv2.addWeighted(img, 1.0, mask, 1.5, 0)
# Display the results
plt.subplot(1, 2, 1)
plt.title('Original')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.subplot()
plt.title('Sharpened')
plt.imshow(cv2.cvtColor(sharpened, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```
